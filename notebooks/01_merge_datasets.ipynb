# 01 — Dataset Preparation and Merging

This notebook prepares the datasets used in the dissertation. It imports five monthly macroeconomic series, standardises their column names and dates, defines the appropriate study periods, and exports the datasets required for the SARIMA and VAR analyses.

No modelling, forecasting or stationarity transformations are performed in this notebook.

**Import libraries**

`pandas` is used to load, transform and merge the datasets. `MonthEnd` is used to represent every monthly observation using the final calendar day of its corresponding month.

```python
import pandas as pd
from pandas.tseries.offsets import MonthEnd
```

**Load the raw datasets**

The five monthly series are imported from CSV files downloaded from the BCRP statistical database. Latin-1 encoding is used because the original files contain Spanish characters, while the first row is skipped because it does not contain the required column headers.

```python
inflation = pd.read_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/raw/inflation.csv", encoding='latin-1', skiprows=1)
interest = pd.read_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/raw/interest_rate.csv", encoding='latin-1', skiprows=1)
exchange = pd.read_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/raw/exchange_rate.csv", encoding='latin-1', skiprows=1)
gdp = pd.read_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/raw/gdp.csv", encoding='latin-1', skiprows=1)
liquidity = pd.read_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/raw/liquidity.csv", encoding='latin-1', skiprows=1)
```

**Standardise column names**

The original BCRP column names are replaced with shorter and consistent English names. Each dataset contains a monthly date column and one variable column.

```python
inflation.columns = ["Date", "Inflation"]
interest.columns = ["Date", "Interest"]
exchange.columns = ["Date", "Exchange Rate"]
gdp.columns = ["Date", "GDP"]
liquidity.columns = ["Date", "Liquidity"]
```

**Standardise the monthly dates**

The original files use Spanish month abbreviations, so these are replaced with English abbreviations before converting the dates to datetime format. All dates are then set to the last day of the month to keep the monthly index consistent across the datasets.

```python
month_map = {
    "Ene": "Jan",
    "Feb": "Feb",
    "Mar": "Mar",
    "Abr": "Apr",
    "May": "May",
    "Jun": "Jun",
    "Jul": "Jul",
    "Ago": "Aug",
    "Sep": "Sep",
    "Oct": "Oct",
    "Nov": "Nov",
    "Dic": "Dec"
}
```

```python
inflation["Date"] = inflation["Date"].replace(month_map, regex=True)
inflation["Date"] = pd.to_datetime(inflation["Date"], format="%b%y")
```

```python
interest["Date"] = interest["Date"].replace(month_map, regex=True)
interest["Date"] = pd.to_datetime(interest["Date"],format="%b%y")
```

```python
exchange["Date"] = exchange["Date"].replace(month_map, regex=True)
exchange["Date"] = pd.to_datetime(exchange["Date"],format="%b%y")
```

```python
gdp["Date"] = gdp["Date"].replace(month_map, regex=True)
gdp["Date"] = pd.to_datetime(gdp["Date"],format="%b%y")
```

```python
liquidity["Date"] = liquidity["Date"].replace(month_map, regex=True)
liquidity["Date"] = pd.to_datetime(liquidity["Date"],format="%b%y")
```

```python
inflation["Date"] = inflation["Date"] + MonthEnd(0)
interest["Date"] = interest["Date"] + MonthEnd(0)
exchange["Date"] = exchange["Date"] + MonthEnd(0)
gdp["Date"] = gdp["Date"] + MonthEnd(0)
liquidity["Date"] = liquidity["Date"] + MonthEnd(0)
```

**Export the full inflation series**

The complete inflation series, covering January 1995 to December 2025, is exported for the univariate SARIMA analysis. This export is completed before restricting the datasets to the shorter multivariate study period.

**Define the multivariate study period**

All five series are restricted to the common period from September 2003 to December 2025. September 2003 is used as the starting point because the BCRP policy interest rate was introduced at that time.

```python
start_date = "2003-09-30"
end_date = "2025-12-31"

inflation = inflation[inflation["Date"].between(start_date, end_date)]
exchange = exchange[exchange["Date"].between(start_date, end_date)]
interest = interest[interest["Date"].between(start_date, end_date)]
gdp = gdp[gdp["Date"].between(start_date, end_date)]
liquidity = liquidity[liquidity["Date"].between(start_date, end_date)]
```

**Export the individual multivariate series**

The five datasets covering the common multivariate study period are exported separately. These files preserve the prepared individual series before they are combined.

```python
inflation.to_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/processed/df_inflation_multivar.csv", index=False)
exchange.to_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/processed/df_exchange.csv", index=False)
interest.to_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/processed/df_interest.csv", index=False)
gdp.to_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/processed/df_gdp.csv", index=False)
liquidity.to_csv("/content/drive/MyDrive/MSc_Dissertation/datasets/processed/df_liquidity.csv", index=False)
```

**Merge the multivariate datasets**

The datasets are merged sequentially using the monthly date column. An inner join retains only dates that are present in every series, producing a complete common dataset for the multivariate analysis.

```python
merged = inflation.merge(gdp, on="Date", how="inner")
merged = merged.merge(exchange, on="Date", how="inner")
merged = merged.merge(interest, on="Date", how="inner")
merged = merged.merge(liquidity, on="Date", how="inner")
```
